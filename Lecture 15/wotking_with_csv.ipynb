# Opening CSV Form Url

```python
import requests
from io import StringIO

url="https://raw.githubusercontent.com/Julien-Marcou/Unicode-Emoji/refs/heads/main/unicode-emoji.csv"
headers={"User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10:14; rv:66.0) Gecko/20100101 Firefox/66.0"}
req=requests.get(url,headers=headers)
data=StringIO(req.text)

pd.read_csv(data)
```

# Converters

```python
def rename(name):
    if(name=='Male'):
        return 'male'
    else:
        return name
```

```python
rename('Male')
```

```python
pd.read_csv('heart_disease_risk_2026.csv',converters={'sex':rename})
```
